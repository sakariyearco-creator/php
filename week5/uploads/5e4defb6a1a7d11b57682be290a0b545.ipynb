## Step 2 · Explain   Answer in a markdown cell. 

### 1. Why `0 or "default"` prints `"default"`
In Python, `0` means **empty or False**, while a non-empty string like `"default"` means **True**. 

The `or` keyword looks for the first item that isn't False. Since `0` is False, Python skips it and picks `"default"`.

---
### 2. Why `-7 // 2` gives `-4` instead of `-3`
The `//` symbol means **round down** to the smaller whole number.

When you divide `-7 / 2`, you get `-3.5`. On a number line, going down (towards lower numbers) from `-3.5` leads to `-4`, not `-3`. 

---
### 3. Why `grade(45)` returns `"A"` & Which lines never run
* **Why it gives "A":** Python reads code line-by-line from top to bottom. The very first line checks `if score >= 0:`. Since `45` is bigger than `0`, Python says "Yes, that's true!" and immediately gives back `"A"` without looking at the rest of the function.
* **Which lines can never run:**
  * `elif score >= 70:`
  * `elif score >= 50:`
  * `return "F"`
  None of these will ever run for any positive score, because `if score >= 0:` catches all positive numbers first.

---
### 4. Why the order of `if` and `elif` matters
Python stops checking conditions the moment it finds one that is `True`. 

When grading scores, you must always check the **highest score first** (like `>= 90`, then `>= 70`, then `>= 50`). If you put `>= 0` at the top, every passing score gets stuck in the first check and lower checks never get a turn.

### Step 3 · Build   Write and test new code. 

```python
def grade(score):
    if score < 0 or score > 100:
        return "Error: Score must be between 0 and 100"   
    if score >= 70:
        return "A"
    elif score >= 60:
        return "B"
    elif score >= 50:
        return "C"
    else:
        return "F"
assert grade(-5) == "Error: Score must be between 0 and 100"
assert grade(0) == "F"
assert grade(49) == "F"
assert grade(50) == "C"
assert grade(69) == "B"
assert grade(70) == "A"
assert grade(100) == "A"
assert grade(101) == "Error: Score must be between 0 and 100"
print("All tests passed successfully!")
```

```python
# Problem demonstration
user_discount = 0
discount = user_discount or 10
print("Buggy result:", discount)  # Prints 10 (Wrong!)

# Fixed version
user_discount = 0
discount = 10 if user_discount is None else user_discount
print("Fixed result:", discount)  # Prints 0 (Correct!)
```
